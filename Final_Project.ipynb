```python
class Combatant:
    def __init__(self, name, hp):
        self.name = name
        self.hp = hp

    def attack(self, other):
        pass  # subclasses define their own attack behavior

    def take_damage(self, amount):
        self.hp -= amount
        if self.hp < 0:
            self.hp = 0

    def is_alive(self):
        return self.hp > 0


class Fairy(Combatant):
    def __init__(self, name, hp=30, power=8):
        super().__init__(name, hp)
        self.power = power

    def attack(self, other):
        if not self.is_alive():
            return
        print(f"{self.name} casts a spell on {other.name} for {self.power} damage!")
        other.take_damage(self.power)


class Human(Combatant):
    def __init__(self, name, hp, power):
        super().__init__(name, hp)
        self.power = power

    def attack(self, other):
        if not self.is_alive():
            return
        print(f"{self.name} attacks {other.name} for {self.power} damage!")
        other.take_damage(self.power)


class Villager(Human):
    def __init__(self, name, hp=10, power=2):
        super().__init__(name, hp, power)

    def attack(self, other):
        if not self.is_alive():
            return
        print(f"{self.name} jabs with a pitchfork at {other.name} for {self.power} damage!")
        other.take_damage(self.power)


class Hunter(Human):
    def __init__(self, name, hp=20, power=5):
        super().__init__(name, hp, power)

    def attack(self, other):
        if not self.is_alive():
            return
        print(f"{self.name} fires an arrow at {other.name} for {self.power} damage!")
        other.take_damage(self.power)


def battle(hero, monsters):
    for monster in monsters:
        print(f"\n--- {hero.name} vs {monster.name} ---")
        while hero.is_alive() and monster.is_alive():
            hero.attack(monster)
            if monster.is_alive():
                monster.attack(hero)
            print(f"{hero.name} HP: {hero.hp} | {monster.name} HP: {monster.hp}")

        if not hero.is_alive():
            print(f"\n{hero.name} has fallen! Game over.")
            return

        print(f"{monster.name} defeated!")

    print(f"\n{hero.name} wins! All humans defeated!")


# Set up the battle
if __name__ == "__main__":
    fairy = Fairy("Luna")
    humans = [Villager("Villager 1"), Villager("Villager 2"), Hunter("Hunter Jake")]
    battle(fairy, humans)
```
