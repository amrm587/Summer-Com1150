```python
# Drake Album Tracker

catalog = {}        # dict: lowercase name -> {"display_name": ..., "year": ..., "rating": ...}
order_added = []    # list: keeps track of order albums were added
collected = set()   # set: unique album names collected (lowercase)

GOAL = 3  # target number of albums to collect
PASSWORD = "drake123"  # Change this to your password


def add_album(name, year, rating):
    """Add an album to the catalog."""
    name_lower = name.lower()  # For lookups

    if name_lower in catalog:
        print(f"'{name}' is already in your collection!")
        return

    # Store both the display name AND the info
    catalog[name_lower] = {"display_name": name, "year": year, "rating": rating}
    order_added.append(name_lower)
    collected.add(name_lower)
    print(f"Added '{name}' ({year}) - Rating: {rating}/10")


def find_album(name):
    """Look up an album by name."""
    name_lower = name.lower()
    album = catalog.get(name_lower)

    if album:
        display = album["display_name"]  # Get the original name
        print(f"{display}: Year {album['year']}, Rating {album['rating']}/10")
    else:
        print(f"'{name}' not found in your collection.")


def list_albums():
    """List all albums in the order they were added."""
    # Ask for password first
    attempt = input("Enter password to view your collection: ")
    if attempt != PASSWORD:
        print("Incorrect password!")
        return

    if not order_added:
        print("Your collection is empty.")
        return

    print("\n--- Your Drake Album Collection ---")
    for name_lower in order_added:
        info = catalog[name_lower]
        display = info["display_name"]  # Get the original name
        print(f"{display} ({info['year']}) - Rating: {info['rating']}/10")
    print(f"Progress: {len(collected)}/{GOAL} albums collected\n")


def menu():
    while True:
        print("\n1. Add album\n2. Look up album\n3. List all albums\n4. Quit")
        choice = input("Choose an option: ")

        if choice == "1":
            name = input("Album name: ")

            # Validate year - must be exactly 4 digits
            while True:
                year = input("Year released (4 digits): ")
                if len(year) == 4 and year.isdigit():
                    break  # Valid year, exit loop
                else:
                    print("Please enter a valid 4-digit year (e.g., 2011)")

            rating = input("Your rating (1-10): ")
            add_album(name, year, rating)

        elif choice == "2":
            name = input("Album name to look up: ")
            find_album(name)

        elif choice == "3":
            list_albums()  # Password check happens inside this function

        elif choice == "4":
            print("Thanks for tracking your Drake albums!")
            break

        else:
            print("Invalid option, try again.")


if __name__ == "__main__":
    menu()
```

2. **claude suggested preserving the original captialization by storing the display name in a seperate funtions so when its time to display the album name even during a look up it will always display the capitalized version of it**

3. **i wanted to set a requirement on how many numbers needed to be entered for the year so i added len(year) == 4**

4. **i set up a password protection for when a user wants to view the whole name**
